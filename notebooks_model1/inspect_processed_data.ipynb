```python
from pathlib import Path
import sys
import pandas as pd
import numpy as np

REPO_ROOT = Path.cwd().parent
if str(REPO_ROOT) not in sys.path:
    sys.path.append(str(REPO_ROOT))

from src.models.input_layer import load_modeling_splits
from src.models.feature_sets import BASE_TABULAR_FEATURES, EXTENDED_TABULAR_FEATURES
```

```python
splits = load_modeling_splits(dataset_name="earthquake_aftershock_v2_gcmt")

train_df = splits["train"]
val_df = splits["val"]
test_df = splits["test"]

print("Train shape:", train_df.shape)
print("Val shape:", val_df.shape)
print("Test shape:", test_df.shape)
```

```python
print("Train columns:")
for c in train_df.columns:
    print(c)
```

```python
possible_label_cols = ["y_24h", "y_72h", "label_24h", "label_72h"]

print("Possible label columns found:")
for c in possible_label_cols:
    if c in train_df.columns:
        print(c)
```

```python
label_cols = [c for c in ["y_24h", "y_72h", "label_24h", "label_72h"] if c in train_df.columns]

for target in label_cols:
    print(f"\nTarget: {target}")
    print("train positive rate:", train_df[target].mean())
    print("val positive rate:", val_df[target].mean())
    print("test positive rate:", test_df[target].mean())
```
