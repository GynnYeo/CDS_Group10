```python
import pandas as pd
import os

# Define the splits directory
splits_dir = '../data/processed/splits/earthquake_aftershock_v2_gcmt'

# Load all splits
train = pd.read_parquet(os.path.join(splits_dir, 'earthquake_aftershock_v2_gcmt__train.parquet'))
val = pd.read_parquet(os.path.join(splits_dir, 'earthquake_aftershock_v2_gcmt__val.parquet'))
test = pd.read_parquet(os.path.join(splits_dir, 'earthquake_aftershock_v2_gcmt__test.parquet'))

# Display split sizes
print("=" * 60)
print("SPLIT SIZES")
print("=" * 60)
splits_info = pd.DataFrame({
    'Split': ['Train', 'Validation', 'Test', 'Total'],
    'Rows': [len(train), len(val), len(test), len(train) + len(val) + len(test)],
    'Columns': [train.shape[1], val.shape[1], test.shape[1], train.shape[1]],
    'Percentage': [
        f"{len(train)/(len(train)+len(val)+len(test))*100:.1f}%",
        f"{len(val)/(len(train)+len(val)+len(test))*100:.1f}%",
        f"{len(test)/(len(train)+len(val)+len(test))*100:.1f}%",
        "100%"
    ]
})
print(splits_info.to_string(index=False))

# Display column names (features and labels)
print("\n" + "=" * 60)
print("FEATURES")
print("=" * 60)
print(f"Total columns: {train.shape[1]}")
print("\nColumn names:")
for col in train.columns:
    print(f"  - {col}")

# Display data types
print("\n" + "=" * 60)
print("DATA TYPES")
print("=" * 60)
print(train.dtypes)

# Display basic statistics for each split
print("\n" + "=" * 60)
print("TRAIN SPLIT STATISTICS")
print("=" * 60)
print(train.describe())

print("\n" + "=" * 60)
print("VALIDATION SPLIT STATISTICS")
print("=" * 60)
print(val.describe())

print("\n" + "=" * 60)
print("TEST SPLIT STATISTICS")
print("=" * 60)
print(test.describe())

# Check for missing values
print("\n" + "=" * 60)
print("MISSING VALUES")
print("=" * 60)
missing_info = pd.DataFrame({
    'Train': train.isnull().sum(),
    'Val': val.isnull().sum(),
    'Test': test.isnull().sum()
})
print(missing_info[missing_info.sum(axis=1) > 0])
if missing_info.sum(axis=1).sum() == 0:
    print("No missing values in any split!")
```
